1. TRAEMOS EL API DE BALOTO PARA PODER CONSUMIR SUS DATOS

```python

# ACTIVIDAD 2 - MODELO OCULTO DE MARKOV
# Fuente: página oficial de resultados de Baloto

!pip -q install requests beautifulsoup4 lxml

import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import re
from bs4 import BeautifulSoup
from io import StringIO

# Página oficial del histórico
URL_BALOTO = "https://www.baloto.com/resultados-baloto"

# Conexión
sesion = requests.Session()

sesion.headers.update({
    "User-Agent": "Mozilla/5.0 (compatible; ProyectoAcademico/1.0)"
})

respuesta = sesion.get(URL_BALOTO, timeout=30)

print("Código de respuesta:", respuesta.status_code)
print("Conexión exitosa:", respuesta.ok)
print("Tamaño de la respuesta:", len(respuesta.text), "caracteres")

if respuesta.ok:
    print("\nConexión con la fuente oficial establecida.")
else:
    raise Exception("No fue posible consultar la página oficial.")

# Inspeccionar las tablas disponibles
tablas = pd.read_html(StringIO(respuesta.text))

print("\nCantidad de tablas encontradas:", len(tablas))

for i, tabla in enumerate(tablas):
    print(f"\nTABLA {i}")
    print("Dimensiones:", tabla.shape)
    display(tabla.head())
```

Identificar automáticamente todas las páginas históricas

```python
# PASO 2: DETECTAR LA PAGINACIÓN COMPLETA

soup = BeautifulSoup(respuesta.text, "lxml")

paginas = []

for enlace in soup.find_all("a", href=True):

    href = enlace["href"]

    coincidencia = re.search(r"[?&]page=(\d+)", href)

    if coincidencia:
        paginas.append(int(coincidencia.group(1)))

# Incluir la primera página
paginas.append(1)

total_paginas = max(paginas)

print("Primera página:", min(paginas))
print("Última página identificada:", total_paginas)
print("Total de páginas:", total_paginas)

# Mostrar enlaces de sorteos de la primera página
print("\nENLACES DE SORTEOS:")

for enlace in soup.find_all("a", href=True):

    href = enlace["href"]

    if re.search(r"/resultados-(baloto|revancha)/\d+", href):
        print(href)
```

Descargar el histórico completo

```python
# PASO 3: DESCARGAR EL HISTÓRICO COMPLETO DE BALOTO Y REVANCHA

import time

registros = []
errores = []

for pagina in range(1, total_paginas + 1):

    url = f"{URL_BALOTO}?page={pagina}"

    try:
        r = sesion.get(url, timeout=30)
        r.raise_for_status()

        soup_pagina = BeautifulSoup(r.text, "lxml")
        tabla_html = soup_pagina.find("table")

        if tabla_html is None:
            errores.append(f"Página {pagina}: tabla no encontrada")
            continue

        filas = tabla_html.find_all("tr")

        for fila in filas[1:]:

            columnas = fila.find_all("td")

            if len(columnas) < 3:
                continue

            fecha = columnas[1].get_text(" ", strip=True)
            resultado = columnas[2].get_text(" ", strip=True)

            enlaces = fila.find_all("a", href=True)

            for enlace in enlaces:

                href = enlace["href"]

                coincidencia = re.search(
                    r"/resultados-(baloto|revancha)/(\d+)",
                    href
                )

                if not coincidencia:
                    continue

                modalidad = coincidencia.group(1).capitalize()
                numero_sorteo = int(coincidencia.group(2))

                numeros = re.findall(r"\d+", resultado)

                if len(numeros) != 6:
                    errores.append(
                        f"Sorteo {numero_sorteo}: resultado inválido"
                    )
                    continue

                numeros = [int(n) for n in numeros]

                balotas = numeros[:5]
                superbalota = numeros[5]

                if (
                    len(set(balotas)) != 5
                    or not all(1 <= n <= 43 for n in balotas)
                    or not 1 <= superbalota <= 16
                ):
                    errores.append(
                        f"Sorteo {numero_sorteo}: números fuera de rango"
                    )
                    continue

                registros.append({
                    "Fecha": fecha,
                    "Sorteo": numero_sorteo,
                    "Modalidad": modalidad,
                    "Balota_1": balotas[0],
                    "Balota_2": balotas[1],
                    "Balota_3": balotas[2],
                    "Balota_4": balotas[3],
                    "Balota_5": balotas[4],
                    "Superbalota": superbalota
                })

        if pagina % 10 == 0:
            print(
                f"Páginas procesadas: {pagina}/{total_paginas} "
                f"| Registros: {len(registros)}"
            )

        time.sleep(0.15)

    except Exception as e:
        errores.append(f"Página {pagina}: {str(e)}")


# Crear la base de datos
df_historico = pd.DataFrame(registros)

# Eliminar posibles duplicados
df_historico = df_historico.drop_duplicates(
    subset=["Sorteo", "Modalidad"]
)

# Organizar los registros
df_historico = df_historico.sort_values(
    by=["Sorteo", "Modalidad"],
    ascending=[False, True]
).reset_index(drop=True)


# Mostrar resultados
print("\n========== RESUMEN DEL HISTÓRICO ==========")

print("Total de registros:", len(df_historico))
print("Sorteos únicos:", df_historico["Sorteo"].nunique())

print("\nRegistros por modalidad:")
print(df_historico["Modalidad"].value_counts())

print("\nErrores encontrados:", len(errores))

print("\nPrimeros registros:")
display(df_historico.head(10))

print("\nÚltimos registros:")
display(df_historico.tail(10))

if errores:
    print("\nPrimeros errores:")
    print(errores[:10])
```

VALIDACION CALIDAD DE LOS DATOS

```python
# PASO 4: VALIDACIÓN DE CALIDAD DE LOS DATOS

# 1. Revisar sorteos con ambas modalidades
modalidades_por_sorteo = (
    df_historico.groupby("Sorteo")["Modalidad"]
    .nunique()
)

sorteos_completos = (modalidades_por_sorteo == 2).all()

# 2. Revisar fechas por sorteo
fechas_por_sorteo = (
    df_historico.groupby("Sorteo")["Fecha"]
    .nunique()
)

fechas_correctas = (fechas_por_sorteo == 1).all()

# 3. Revisar números principales sin repeticiones
columnas_balotas = [
    "Balota_1", "Balota_2", "Balota_3",
    "Balota_4", "Balota_5"
]

sin_repetidos = df_historico.apply(
    lambda fila: len(set(fila[columnas_balotas])) == 5,
    axis=1
).all()

# 4. Revisar rangos
rango_balotas = df_historico[columnas_balotas].apply(
    lambda columna: columna.between(1, 43).all()
).all()

rango_superbalota = (
    df_historico["Superbalota"].between(1, 16).all()
)

# 5. Revisar duplicados
sin_duplicados = not df_historico.duplicated(
    subset=["Sorteo", "Modalidad"]
).any()

# Resultados
validaciones = {
    "Dos modalidades por sorteo": sorteos_completos,
    "Fechas coincidentes": fechas_correctas,
    "Cinco balotas sin repetición": sin_repetidos,
    "Balotas entre 1 y 43": rango_balotas,
    "Superbalota entre 1 y 16": rango_superbalota,
    "Sin registros duplicados": sin_duplicados
}

for nombre, resultado in validaciones.items():
    print(f"{nombre}: {'CORRECTO' if resultado else 'REVISAR'}")

print("\nVALIDACIÓN GENERAL:",
      "APROBADA" if all(validaciones.values()) else "REQUIERE REVISIÓN")
```

```python
# PASO 5: EXPORTAR DATOS Y CALCULAR FRECUENCIAS

from google.colab import files

# Columnas de las cinco balotas principales
columnas_balotas = [
    "Balota_1", "Balota_2", "Balota_3",
    "Balota_4", "Balota_5"
]

# Crear tablas de frecuencias
frecuencias_principales = []
frecuencias_superbalota = []

for modalidad in ["Baloto", "Revancha"]:

    datos = df_historico[
        df_historico["Modalidad"] == modalidad
    ]

    total_sorteos = len(datos)

    # Frecuencia de balotas principales
    valores = datos[columnas_balotas].values.flatten()

    conteo = pd.Series(valores).value_counts()

    for numero in range(1, 44):

        apariciones = int(conteo.get(numero, 0))

        frecuencias_principales.append({
            "Modalidad": modalidad,
            "Numero": numero,
            "Apariciones": apariciones,
            "Probabilidad_empirica_pct": round(
                apariciones / total_sorteos * 100, 2
            )
        })

    # Frecuencia de Superbalota
    conteo_sb = datos["Superbalota"].value_counts()

    for numero in range(1, 17):

        apariciones = int(conteo_sb.get(numero, 0))

        frecuencias_superbalota.append({
            "Modalidad": modalidad,
            "Superbalota": numero,
            "Apariciones": apariciones,
            "Probabilidad_empirica_pct": round(
                apariciones / total_sorteos * 100, 2
            )
        })


df_frecuencias = pd.DataFrame(frecuencias_principales)
df_frecuencias_sb = pd.DataFrame(frecuencias_superbalota)

# Ordenar de mayor a menor frecuencia
df_frecuencias = df_frecuencias.sort_values(
    ["Modalidad", "Apariciones"],
    ascending=[True, False]
)

df_frecuencias_sb = df_frecuencias_sb.sort_values(
    ["Modalidad", "Apariciones"],
    ascending=[True, False]
)

# Mostrar resultados
print("FRECUENCIAS DE BALOTAS PRINCIPALES")
display(df_frecuencias)

print("FRECUENCIAS DE SUPERBALOTA")
display(df_frecuencias_sb)

# Exportar a Excel con varias hojas
with pd.ExcelWriter("Historico_Baloto_Revancha.xlsx") as writer:

    df_historico.to_excel(
        writer, sheet_name="Historico", index=False
    )

    df_frecuencias.to_excel(
        writer, sheet_name="Frecuencias_Balotas", index=False
    )

    df_frecuencias_sb.to_excel(
        writer, sheet_name="Frecuencias_Superbalota", index=False
    )

# Exportar también CSV
df_historico.to_csv(
    "Historico_Baloto_Revancha.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\nArchivos generados correctamente.")

files.download("Historico_Baloto_Revancha.xlsx")
```

 CONSTRUIR LAS OBSERVACIONES DEL HMM

```python
# PASO 7: CONSTRUIR LAS OBSERVACIONES DEL HMM

columnas_balotas = [
    "Balota_1", "Balota_2", "Balota_3",
    "Balota_4", "Balota_5"
]

# Diccionario para guardar las secuencias
secuencias_hmm = {}

# Identificar los cinco números más frecuentes por modalidad
numeros_frecuentes = {}

for modalidad in ["Baloto", "Revancha"]:

    frecuencias = df_frecuencias[
        df_frecuencias["Modalidad"] == modalidad
    ]

    top_5 = (
        frecuencias.nlargest(5, "Apariciones")["Numero"]
        .tolist()
    )

    numeros_frecuentes[modalidad] = top_5

    # Filtrar y ordenar cronológicamente
    datos = df_historico[
        df_historico["Modalidad"] == modalidad
    ].copy()

    datos = datos.sort_values("Sorteo").reset_index(drop=True)

    # Contar cuántos números frecuentes aparecen en cada sorteo
    observaciones = []

    for _, fila in datos.iterrows():

        numeros_sorteo = set(
            fila[columnas_balotas].tolist()
        )

        cantidad = len(
            numeros_sorteo.intersection(set(top_5))
        )

        observaciones.append(cantidad)

    datos["Observacion"] = observaciones

    secuencias_hmm[modalidad] = datos

    print("\n", "=" * 45)
    print("MODALIDAD:", modalidad)
    print("Cinco números frecuentes:", top_5)

    print("\nDistribución de observaciones:")
    print(datos["Observacion"].value_counts().sort_index())

    print("\nPrimeros sorteos cronológicos:")
    display(
        datos[["Fecha", "Sorteo", "Observacion"]].head(10)
    )

    print("\nTotal de observaciones:", len(observaciones))
```

```python
# PASO 8: RESUMEN ESTADÍSTICO DE LAS OBSERVACIONES

for modalidad in ["Baloto", "Revancha"]:

    datos = secuencias_hmm[modalidad]

    print("\n" + "=" * 50)
    print("MODALIDAD:", modalidad)

    distribucion = (
        datos["Observacion"]
        .value_counts()
        .reindex(range(6), fill_value=0)
        .sort_index()
    )

    resumen = pd.DataFrame({
        "Cantidad_sorteos": distribucion,
        "Porcentaje": (
            distribucion / len(datos) * 100
        ).round(2)
    })

    display(resumen)

    print("Total de observaciones:", len(datos))
    print("Promedio de números frecuentes por sorteo:",
          round(datos["Observacion"].mean(), 4))

    print("Valor mínimo:", datos["Observacion"].min())
    print("Valor máximo:", datos["Observacion"].max())

    # Preparar secuencia numérica
    secuencia = datos["Observacion"].to_numpy(dtype=int)

    print("Secuencia válida:",
          bool(((secuencia >= 0) & (secuencia <= 5)).all()))
```

ENTRENAMIENTO DEL MODELO HMM

```python
# PASO 9: ENTRENAMIENTO DEL MODELO HMM

!pip -q install hmmlearn

from hmmlearn.hmm import CategoricalHMM
import numpy as np

modelos_hmm = {}
resultados_hmm = {}

for modalidad in ["Baloto", "Revancha"]:

    print("\n" + "=" * 55)
    print("ENTRENANDO MODELO:", modalidad)

    # Obtener secuencia cronológica
    secuencia = (
        secuencias_hmm[modalidad]["Observacion"]
        .to_numpy(dtype=int)
        .reshape(-1, 1)
    )

    mejor_modelo = None
    mejor_puntuacion = -np.inf

    # Probar diez inicializaciones
    for semilla in range(10):

        modelo = CategoricalHMM(
            n_components=3,
            n_iter=300,
            tol=0.0001,
            random_state=semilla,
            verbose=False,
            init_params="ste",
            params="ste"
        )

        modelo.n_features = 6

        try:
            modelo.fit(secuencia)

            puntuacion = modelo.score(secuencia)

            if puntuacion > mejor_puntuacion:
                mejor_puntuacion = puntuacion
                mejor_modelo = modelo

        except Exception as error:
            print("Error en inicialización:", semilla, error)

    if mejor_modelo is None:
        raise Exception(
            f"No fue posible entrenar el modelo {modalidad}"
        )

    modelos_hmm[modalidad] = mejor_modelo

    resultados_hmm[modalidad] = {
        "Log_verosimilitud": mejor_puntuacion,
        "Probabilidad_inicial": mejor_modelo.startprob_,
        "Matriz_transicion": mejor_modelo.transmat_,
        "Matriz_emision": mejor_modelo.emissionprob_
    }

    print("Log-verosimilitud:", round(mejor_puntuacion, 4))

    print("\nProbabilidad inicial:")
    display(pd.DataFrame(
        mejor_modelo.startprob_.reshape(1, -1),
        columns=["Estado 0", "Estado 1", "Estado 2"]
    ))

    print("\nMatriz de transición:")
    display(pd.DataFrame(
        mejor_modelo.transmat_,
        index=["Estado 0", "Estado 1", "Estado 2"],
        columns=["Estado 0", "Estado 1", "Estado 2"]
    ))

    print("\nMatriz de emisión:")
    display(pd.DataFrame(
        mejor_modelo.emissionprob_,
        index=["Estado 0", "Estado 1", "Estado 2"],
        columns=[f"Observación {i}" for i in range(6)]
    ))

print("\nENTRENAMIENTO FINALIZADO")
```

 VALIDACIÓN E INTERPRETACIÓN DE LOS ESTADOS

```python
# PASO 10: VALIDACIÓN E INTERPRETACIÓN DE LOS ESTADOS

estados_ocultos = {}

for modalidad in ["Baloto", "Revancha"]:

    modelo = modelos_hmm[modalidad]

    secuencia = (
        secuencias_hmm[modalidad]["Observacion"]
        .to_numpy(dtype=int)
        .reshape(-1, 1)
    )

    # Obtener estados ocultos mediante Viterbi
    estados = modelo.predict(secuencia)

    estados_ocultos[modalidad] = estados

    print("\n" + "=" * 60)
    print("MODALIDAD:", modalidad)

    # Comprobar normalización
    print("\nSuma de probabilidades iniciales:")
    print(round(modelo.startprob_.sum(), 6))

    print("\nSuma de cada fila de transición:")
    print(modelo.transmat_.sum(axis=1).round(6))

    print("\nSuma de cada fila de emisión:")
    print(modelo.emissionprob_.sum(axis=1).round(6))

    # Comprobar convergencia
    print("\n¿Convergió el algoritmo?")
    print(modelo.monitor_.converged)

    print("Iteraciones realizadas:")
    print(modelo.monitor_.iter)

    # Distribución de estados
    print("\nCantidad de sorteos por estado:")
    print(pd.Series(estados).value_counts().sort_index())

    # Interpretar estados según la observación esperada
    observaciones = np.arange(6)

    medias = modelo.emissionprob_ @ observaciones

    interpretacion = pd.DataFrame({
        "Estado": range(3),
        "Observacion_esperada": medias
    })

    interpretacion["Nivel"] = pd.cut(
        interpretacion["Observacion_esperada"],
        bins=[-0.01, 0.75, 1.5, 5],
        labels=["Bajo", "Medio", "Alto"]
    )

    print("\nInterpretación de los estados:")
    display(interpretacion)

    # Agregar estados a los datos
    secuencias_hmm[modalidad]["Estado_oculto"] = estados

print("\nVALIDACIÓN TERMINADA")
```

EVALUACIÓN DEL HMM FRENTE A UN MODELO BASE

```python
# PASO 12: EVALUACIÓN DEL HMM FRENTE A UN MODELO BASE

import numpy as np
import pandas as pd

resultados_evaluacion = []

for modalidad in ["Baloto", "Revancha"]:

    secuencia = (
        secuencias_hmm[modalidad]["Observacion"]
        .to_numpy(dtype=int)
    )

    modelo = modelos_hmm[modalidad]

    n = len(secuencia)

    # Log-verosimilitud del HMM
    loglik_hmm = modelo.score(
        secuencia.reshape(-1, 1)
    )

    # Modelo base: observaciones independientes
    frecuencias = (
        pd.Series(secuencia)
        .value_counts(normalize=True)
        .reindex(range(6), fill_value=0)
        .to_numpy()
    )

    probabilidades = frecuencias[secuencia]

    loglik_base = np.sum(np.log(probabilidades))

    # Cantidad de parámetros libres
    parametros_hmm = (
        2 + 6 + 15
    )

    parametros_base = 5

    # Criterio de información bayesiano
    bic_hmm = (
        -2 * loglik_hmm
        + parametros_hmm * np.log(n)
    )

    bic_base = (
        -2 * loglik_base
        + parametros_base * np.log(n)
    )

    resultados_evaluacion.append({
        "Modalidad": modalidad,
        "Loglik_HMM": loglik_hmm,
        "Loglik_Base": loglik_base,
        "BIC_HMM": bic_hmm,
        "BIC_Base": bic_base,
        "Diferencia_BIC": bic_hmm - bic_base
    })

df_evaluacion = pd.DataFrame(resultados_evaluacion)

display(df_evaluacion.round(4))

print("\nINTERPRETACIÓN:")

for _, fila in df_evaluacion.iterrows():

    if fila["Diferencia_BIC"] < 0:
        conclusion = "El HMM obtiene menor BIC."
    else:
        conclusion = "El modelo base obtiene menor BIC."

    print(f"{fila['Modalidad']}: {conclusion}")
```

REENTRENAMIENTO DE BALOTO

```python
# PASO 13: REENTRENAMIENTO DE BALOTO

secuencia_baloto = (
    secuencias_hmm["Baloto"]["Observacion"]
    .to_numpy(dtype=int)
    .reshape(-1, 1)
)

mejor_modelo_baloto = None
mejor_puntuacion_baloto = -np.inf

for semilla in range(20):

    modelo = CategoricalHMM(
        n_components=3,
        n_iter=2000,
        tol=0.00001,
        random_state=semilla,
        verbose=False,
        init_params="ste",
        params="ste"
    )

    modelo.n_features = 6
    modelo.fit(secuencia_baloto)

    puntuacion = modelo.score(secuencia_baloto)

    if puntuacion > mejor_puntuacion_baloto:
        mejor_puntuacion_baloto = puntuacion
        mejor_modelo_baloto = modelo

# Actualizar el modelo anterior
modelos_hmm["Baloto"] = mejor_modelo_baloto

# Actualizar resultados
estados_ocultos["Baloto"] = mejor_modelo_baloto.predict(
    secuencia_baloto
)

secuencias_hmm["Baloto"]["Estado_oculto"] = (
    estados_ocultos["Baloto"]
)

print("NUEVO ENTRENAMIENTO DE BALOTO")
print("Log-verosimilitud:", mejor_puntuacion_baloto)
print("Iteraciones:", mejor_modelo_baloto.monitor_.iter)
print("Convergencia reportada:", mejor_modelo_baloto.monitor_.converged)

print("\nÚltimas diferencias:")
print(
    np.diff(mejor_modelo_baloto.monitor_.history)[-10:].round(8)
)

print("\nProbabilidad inicial:")
display(pd.DataFrame(
    [mejor_modelo_baloto.startprob_],
    columns=["Estado 0", "Estado 1", "Estado 2"]
))

print("\nMatriz de transición:")
display(pd.DataFrame(mejor_modelo_baloto.transmat_))

print("\nMatriz de emisión:")
display(pd.DataFrame(mejor_modelo_baloto.emissionprob_))
```

ACTUALIZAR ESTADOS Y EVALUACIÓN BIC

```python
# PASO 14: ACTUALIZAR ESTADOS Y EVALUACIÓN BIC

resultados_evaluacion = []

for modalidad in ["Baloto", "Revancha"]:

    modelo = modelos_hmm[modalidad]

    secuencia = (
        secuencias_hmm[modalidad]["Observacion"]
        .to_numpy(dtype=int)
        .reshape(-1, 1)
    )

    # Actualizar estados ocultos
    estados = modelo.predict(secuencia)

    estados_ocultos[modalidad] = estados

    secuencias_hmm[modalidad]["Estado_oculto"] = estados

    # Log-verosimilitud del HMM
    n = len(secuencia)
    loglik_hmm = modelo.score(secuencia)

    # Modelo base independiente
    frecuencias = (
        pd.Series(secuencia.flatten())
        .value_counts(normalize=True)
        .reindex(range(6), fill_value=0)
        .to_numpy()
    )

    loglik_base = np.sum(
        np.log(frecuencias[secuencia.flatten()])
    )

    # Parámetros libres
    parametros_hmm = 23
    parametros_base = 5

    bic_hmm = (
        -2 * loglik_hmm
        + parametros_hmm * np.log(n)
    )

    bic_base = (
        -2 * loglik_base
        + parametros_base * np.log(n)
    )

    resultados_evaluacion.append({
        "Modalidad": modalidad,
        "Loglik_HMM": loglik_hmm,
        "Loglik_Base": loglik_base,
        "BIC_HMM": bic_hmm,
        "BIC_Base": bic_base,
        "Diferencia_BIC": bic_hmm - bic_base,
        "Iteraciones": modelo.monitor_.iter,
        "Convergencia": modelo.monitor_.converged
    })

df_evaluacion = pd.DataFrame(resultados_evaluacion)

print("EVALUACIÓN ACTUALIZADA")
display(df_evaluacion.round(5))

for modalidad in ["Baloto", "Revancha"]:

    print("\nMODALIDAD:", modalidad)

    estados = estados_ocultos[modalidad]

    print("Distribución de estados:")
    display(
        pd.Series(estados)
        .value_counts()
        .sort_index()
        .rename("Cantidad de sorteos")
    )
```

```python
# PASO 15: EXPORTACIÓN DEFINITIVA DE RESULTADOS

with pd.ExcelWriter(
    "Resultados_Finales_HMM_Baloto_Revancha.xlsx"
) as writer:

    # Evaluación actualizada
    df_evaluacion.to_excel(
        writer,
        sheet_name="Evaluacion_BIC",
        index=False
    )

    for modalidad in ["Baloto", "Revancha"]:

        modelo = modelos_hmm[modalidad]

        # Probabilidades iniciales
        pd.DataFrame({
            "Estado": ["Estado 0", "Estado 1", "Estado 2"],
            "Probabilidad": modelo.startprob_
        }).to_excel(
            writer,
            sheet_name=f"{modalidad}_Inicial",
            index=False
        )

        # Transición
        pd.DataFrame(
            modelo.transmat_,
            index=["Estado 0", "Estado 1", "Estado 2"],
            columns=["Estado 0", "Estado 1", "Estado 2"]
        ).to_excel(
            writer,
            sheet_name=f"{modalidad}_Transicion"
        )

        # Emisión
        pd.DataFrame(
            modelo.emissionprob_,
            index=["Estado 0", "Estado 1", "Estado 2"],
            columns=[f"Observacion_{i}" for i in range(6)]
        ).to_excel(
            writer,
            sheet_name=f"{modalidad}_Emision"
        )

        # Secuencia histórica y estados ocultos
        secuencias_hmm[modalidad].to_excel(
            writer,
            sheet_name=f"{modalidad}_Secuencia",
            index=False
        )

print("Excel definitivo actualizado correctamente.")

from google.colab import files

files.download(
    "Resultados_Finales_HMM_Baloto_Revancha.xlsx"
)
```
